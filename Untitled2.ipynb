```python
a=np.random.randint(2,15,6)
b=np.random.randint(2,20,6)

print(a)
# [14 14  4 12  6  5]

print(b)
# [13  4 15 11 13 17]

a=a.reshape((2,3))
b=b.reshape((2,3))
print(a)
# [[14 14  4]
#  [12  6  5]]

print(b)
# [[13  4 15]
#  [11 13 17]]



print(np.concatenate((a,b)))#pass as a tuple
#[[14 14  4]
#  [12  6  5]
#  [13  4 15]
#  [11 13 17]]


print(np.concatenate((a,b),axis=1))
# [[14 14  4 13  4 15]
#  [12  6  5 11 13 17]]
```

```python
sc=[]
for i in range(7):
  x=int(input())
  sc.append(x)

scores=np.array(sc)

print(scores)


days=np.array(['monday','tuesday','wednesday','thursday','friday','saturday','sunday'])


for i in range(7):

  print(f'on {days[i]} he scored : {scores[i]}')


print(f'the minimum score of player is {min(scores)}')
print(f'the maximum score of player is {max(scores)}')
print(f'the average score of player is {np.mean(scores)}')
print(f'the total score of player is {np.sum(scores)}')
print(f'the median score of player is {np.median(scores)}')
print(f'the standard deviation of player is {np.std(scores)}')
```
