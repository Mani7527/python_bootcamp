```python
import numpy as np
n=int(input("Enter number of items: "))
lst=[]
ITEMS=['Item 1','Item 2','Item 3','Item 4','Item 5','Item 6','Item 7']
for i in range(n):

  price=int(input(f"Enter price for {ITEMS[i]}:\n"))
  lst.append(price)
# print(" prices:", lst)


item_prices = np.array(lst)
print("Item prices ", item_prices)

std_price=np.std(item_prices)
print("SD of the all prices :", std_price)

max_price=np.max(item_prices)
max_price_index = np.argmax(item_prices)
max_price_item = ITEMS[max_price_index]
print(f"Max price is for {max_price_item}: {max_price}")

min_price=np.min(item_prices)
min_price_index=np.argmin(item_prices)
min_price_item = ITEMS[min_price_index]
print(f"Min price is for {min_price_item}: {min_price}")

mean_price=np.mean(item_prices)
print("Mean of prices : ",mean_price)

median_price=np.median(item_prices)

if median_price in item_prices:
    median_price_index = np.where(item_prices == median_price)[0][0]
    median_price_item = ITEMS[median_price_index]
    print(f"Median price is for {median_price_item}: {median_price}")
else:
    print(f"Median price: {median_price} (This value is not present )")
```

```python
#generate a 5*5 matrix of random int from(1-90)
a=np.random.randint(1,90,25)
print("a 1d matrix of random elements:")
print(a)
a=a.reshape((5,5))
print("arraay of 5 cross 5:")
print(a)
#print mean row and column wise
print("mean of row :")
row_mean=np.mean(a,axis=1)
print(row_mean)
# print(np.mean(),axis=1)
print("mean of column :")
col_mean=np.mean(a,axis=0)
print(col_mean)




#replace even with -1
# a[a%2==0]=-1

print("after replacing the even number with -1:")
print(np.where(a%2==0,-1,a))

#sort the array and column wise
print("sort the array column wise:")
print(np.sort(a,axis:=0))
print("sort the array row wise:")
print(np.sort(a,axis:=1))


#find sum row and column wise
print("sum of array column wise:")
print(np.sum(a,axis:=0))
print("sum of array row wise:")
print(np.sum(a,axis:=1))

#count the no of negative in each column

print(np.sum(np.where(a<0,1,0),axis:=0))



```

```python
#inventory control system
#task:
#create a 5*5 array(stock levels)
#create another 5*5 array (demand)
#compute:
#remaining stock=stock-demand
#replace:
#negative values ->0(out of stock)
#find:
#total shortage
#products with zero stock
#sort each row based on remaining stock


stock_level=np.random.randint(20,82,25)
stock_level=stock_level.reshape((5,5))
print("stock_level is:")
print(stock_level)

demand=np.random.randint(20,81,25)
demand=demand.reshape((5,5))
print("demand is:")
print(demand)

remaining_stock=stock_level-demand
print("reamining stock is:")
print(remaining_stock)

modified_stock=np.where(remaining_stock<0,0,remaining_stock)
print("modified stock is:")
print(modified_stock)

# total_shoratge=np.sum(np.where(remaining_stock<0,remaining_stock,0))
# if total_shoratge <0:
#   total_shoratge=total_shoratge*-1
# print("total shoratge is :")
#print(total_shoratge)

total_shortage=np.sum(np.where(remaining_stock<0,remaining_stock*-1,0),axis=0)
print("total shortage is :")


print(total_shortage)

products_zerostock=np.sum(np.where(modified_stock<0,1,0),axis=0)
print("products with zero stock is:")
print(products_zerostock)

print("sort each row based on remaining stock:")

print(np.sort(modified_stock,axis=1))

```
